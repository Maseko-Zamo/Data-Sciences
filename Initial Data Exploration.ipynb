```python
%pip install matplotlib
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
```

```python
#Attrition Bar
df = pd.read_csv("HR Employee Attrition.csv")

#Atterition Bar graph
Yes_Att = df["Attrition"].value_counts()["Yes"]
No_Att = df["Attrition"].value_counts()["No"]

plt.bar(["Yes","No"], [Yes_Att,No_Att],color=["Blue","Black"])

plt.title("Attrition")
plt.xlabel("YES / NO")
plt.ylabel("Number of Employees")

plt.show()

print(f" Interpretation \n The number of employees who left are well over 200 , theres is  more than 1200 people who stay within thge company which is a good attrition ratio")


#Attrition Pie
att_c = df["Attrition"].value_counts()

plt.pie(att_c,
labels=att_c.index,
colors=["Black","Blue"],
autopct="%1.1f%%")

plt.title("Employee Attrition (Turnover)")
plt.show()

print(f"Interpretation \n The percentage of the people who leave are 16.1% meaning the company has a turnover")
```

```python

# EDA - Data Exploration

#Graph 2 Attrition vs Overtime
plt.subplot(1, 2, 1)
O_Att = df[df["OverTime"] == "No"]["Attrition"].value_counts()

O_Att.plot(kind="bar",color=["Black","Blue"])
plt.title("Attrition of people with NO overtime")
plt.xlabel("Attriton")
plt.ylabel("Number of Employees")

plt.subplot(1, 2, 2)

O_Att = df[df["OverTime"] == "Yes"]["Attrition"].value_counts()

O_Att.plot(kind="bar",color=["Black","Blue"])
plt.title("Attrition of people with overtime")
plt.xlabel("Attrition")
plt.ylabel("Number of Employees")
plt.subplots_adjust(wspace=1.5)
plt.show()

print(f" Interpretation \n The number of people who take overtime and the number of people who do not take overtime and leave the companyb are roughly the same")



#Attrition by department
Dep1 = df[df["Attrition"] == "Yes"]["Department"].value_counts()

Dep1.plot(kind="bar",color=["Black","Blue","Green"])
plt.title("Attrition by Department")
plt.xlabel("Departments")
plt.ylabel("Number of Employees who left")
plt.show()

print(f" Interpretation \n The Department with the highest attrition is the Research and Development department them follows the sales department")
```

```python
#Grouping the age in different age groups
bins = np.linspace(df["Age"].min(), df["Age"].max(), 5)
grouped_ages = ["18-28.5","28.5-39","39-49.5","49.5-60"]
df["Age_Grouped"] = pd.cut(
    df["Age"],
    bins=bins,
    labels=grouped_ages,
    include_lowest=True
)
df["Age_Grouped"] = df["Age_Grouped"]


Age_Att = df[df["Attrition"] == "Yes"]["Age_Grouped"].value_counts().sort_index()

Age_Att.plot(kind="bar", color=["Blue","Black","Green","Yellow"])
plt.title("Attrition by Age Group")
plt.xlabel("Age Groups")
plt.ylabel("Number of Employees")
plt.show()

print(f" Age By Attrition \n There graph and data seems to show the age group 28-39 being the highest Age Group who are leaving the buiness with the second best being the age group 18 to 28, Essentially the early aged employees are more likley to leave the company as compared to the older employees who are between 39 and 60")


#Attrition by department
Dep1 = df[df["Attrition"] == "Yes"]["Department"].value_counts()

Dep1.plot(kind="bar",color=["Black","Blue","Green"])
plt.title("Attrition by Department")
plt.xlabel("Departments")
plt.ylabel("Number of Employees who left")
plt.show()

print(f" Interpretation \n The Department with the highest attrition is the Research and Development department them follows the sales department")
```

```python
# Histogram 1 - Age distribution
df["Age"].hist()

plt.title("Age Distribution")
plt.xlabel("Age")
plt.ylabel("Frequency")
plt.show()

print(f" Age Distribution Interpretation \n The business has more peoplebetween teh ages of 28.5 to 39 this may be an issue given the fact that the data says that there is more people who leave within this group, the second highest humber of emoployees present in the business is Age group 39 - 49.5")

# Histogram 2 - Monthly Income
df["Department"].hist()

plt.title("Department Distribution")
plt.xlabel("Department")
plt.ylabel("Frequency")
plt.show()

print(f" Department Distribution \n The Research & Development Department has more employees than the other departments than comes sales and lastly HR, it is to be noted that this is the same department with more people leaving")

```
